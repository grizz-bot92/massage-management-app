```python
import pandas as pd
```

```python
df_2021 = pd.read_csv('../../backend/data/massage2021.csv')
df_2022 = pd.read_csv('../../backend/data/massage2022.csv')
df_2023 = pd.read_csv('../../backend/data/massage2023.csv')
df_2024 = pd.read_csv('../../backend/data/massage2024.csv')
```
