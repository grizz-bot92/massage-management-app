```python
import pandas as pd
```

```python
df_2021 = pd.read_csv('../../backend/data/massage2021.csv')
df_2022 = pd.read_csv('../../backend/data/massage2022.csv')
df_2023 = pd.read_csv('../../backend/data/massage2023.csv')
df_2024 = pd.read_csv('../../backend/data/massage2024.csv')
```

```python
df_2021.head()
df_2021.info()
df_2021['Status'].value_counts()
```

```python
df = pd.concat([df_2021, df_2022, df_2023, df_2024], ignore_index=True)
print(df.shape)
df['Status'].value_counts()
```

```python
status_map = {
  'Paid/Complete': 'completed',
  'Cancelled': 'cancelled',
  'No Show': 'no_show',
  'Confirmed': 'completed',
  'Confirmed': 'completed',
  'Guest Checked-in (In Service)': 'completed',
  'Guest Checked-in': 'completed'    
}

df['status'] = df['Status'].map(status_map)
df['status'].value_counts()
```

```python
df['appointment_date'] = pd.to_datetime(df['Appointment On'], format='mixed')
df['appointment_date'].head()
```
